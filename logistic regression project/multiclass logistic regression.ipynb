```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sea
%matplotlib inline 
```

```python
from sklearn.datasets import make_classification
a,b=make_classification(n_samples=1000,n_classes=3,n_features=10,random_state=42,n_informative=4)
```

```python
b
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(a,b,random_state=42,test_size=0.30
                                               )
```

```python
x_train.shape,y_train.shape,x_test.shape,y_test.shape
```

```python
from sklearn.linear_model import LogisticRegression
logistic=LogisticRegression(multi_class='ovr')
```

```python
logistic.fit(x_train,y_train)
```

IMPLEMENTATION OF HYPERPARAMETER TUNING

```python
### GRIDSEARCHCV
model=LogisticRegression()
from sklearn.model_selection import StratifiedKFold
cv=StratifiedKFold()
penalty=['l1','l2','elasticnet']
c_values=[100,10,1.5,1.7,0.01,0.1,0.01]
solver=['newton-cg','lbfgs','sag','saga','liblinear']
```

```python
params=dict(penalty=penalty,C=c_values,solver=solver)
```

```python
from sklearn.model_selection import GridSearchCV
grid=GridSearchCV(estimator=model,param_grid=params,scoring='accuracy',n_jobs=-1,cv=cv)
```

```python
grid.fit(x_train,y_train)
```

```python
grid.best_score_,grid.best_params_
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,classification_report
```

```python
y_pred=grid.predict(x_test)
precision_score(y_test,y_pred,average='macro')
```

```python
print('accuracy is ',accuracy_score(y_test,grid.predict(x_test)))
print('precision_score is ',precision_score(y_test,y_pred,average='micro'))
print('recall ',recall_score(y_test,grid.predict(x_test),average='micro'))
print('the classification report  is ',classification_report(y_test,grid.predict(x_test)))
```

```python
from sklearn.model_selection import RandomizedSearchCV
random=RandomizedSearchCV(estimator=model,param_distributions=params,scoring='accuracy',n_jobs=-1,cv=cv)
```

```python
random.fit(x_train,y_train)
random.best_params_,random.best_score_
```
